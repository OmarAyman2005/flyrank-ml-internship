## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
declining_share = declining_pages / total_pages

high_impression_declining = (
    (df["trend_direction"] == "down") &
    (df["impressions_90d"] >= 100)
).sum()

print(f"Total pages: {total_pages:,}")
print(f"Declining pages: {declining_pages:,} ({declining_share:.1%})")
print(f"Declining pages with >=100 impressions: {high_impression_declining:,}")
```
